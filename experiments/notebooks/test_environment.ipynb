### PARAMS IV

```python
seed = 42
environment = "iv"
reward_node = "Y"
atomic = False
include_empty = False
p_x = 0.6
p_z =0.7
p_y =  0.9
p_zy = 0.5
```

```python
env_params = {
    "environment": environment,
    "reward_node": reward_node,
    "atomic": atomic,
    "include_empty": include_empty,
    "p_x": p_x,
    "p_z": p_z,
    "p_y": p_y,
    "p_zy": p_zy,
}
```

```python
agent_names = ["POMIS-UCB", "PHT-UCB-global", "PHT-UCB-arm", "SR-UCB"]
agent_params = {
    "c": 1.41421356237,
    "delta": 0.005,
    "lambda": 20.0,
    "min_samples_for_detection": 30,
    "atomic": atomic,
}
```

```python
schedule = StationarySchedule()

env = build_environment(env_params, seed, schedule=StationarySchedule())
reward_node = env.reward_node

agents = build_agents(agent_names,  agent_params, env)

action_space: set[Intervention] = set(env.action_space)
agent_action_space = {arm for agent in agents.values() for arm in agent.arms}
effective_action_space = action_space & agent_action_space

for action in effective_action_space:
    print(
        f"Arm: {action}, Expected reward: "
        f"{env.scm.expected_value_binary(variable=reward_node, intervention=action)}"
    )
```
