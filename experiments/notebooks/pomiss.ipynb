```python
from cmab.scm.causal_diagram import CausalDiagram
from cmab.algorithms.pomis.pomis_sets import POMISs

from npsem.model import CausalDiagram as NPSEM_CausalDiagram
from npsem.where_do import POMISs as NPSEM_POMISs

nodes = ["S", "T", "W", "Z", "X", "Y"]

directed_edges = [
("S", "W"),
("T", "X"),
("T", "Y"),
("W", "Y"),
("Z", "X"),
("X", "Y"),]

bidirected_edges = [("Z", "Y", "U_ZY"), ("W", "T", "U_TW")]

target="Y"


my_cd = CausalDiagram(
    nodes=nodes,  
    directed_edges=directed_edges,
    bidirected_edges=bidirected_edges,
)

npsem_cd = NPSEM_CausalDiagram(
    vs=nodes, 
    directed_edges=directed_edges,
    bidirected_edges=bidirected_edges,
)


pomis_mine = POMISs(G=my_cd, Y=target)
pomis_npsem = NPSEM_POMISs(G=npsem_cd, Y=target)
```

```python
pomis_mine == pomis_npsem
```
